```python
# Keep appending the code generated to this cell, or add more cells below this to execute in parts
import pandas as pd

df = pd.read_csv("dataset.csv", header=0)

df.head()
```

```python
print(df.info())
print("\nColumn data types:")
df.dtypes
```

```python
df["Lower Confidence Interval"]=df["Lower Confidence Interval"].astype("string")
import numpy as np

cols = ["Health (Life Expectancy)",
        "Economy (GDP per Capita)",
        "Lower Confidence Interval",
        "Upper Confidence Interval"]

df[cols] = df[cols].apply(lambda x: x.str.strip().replace('"', np.nan))
```

```python
columns = ["Lower Confidence Interval","Upper Confidence Interval", "Economy (GDP per Capita)", "Health (Life Expectancy)","Freedom"]

for column in columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")
df.dtypes
```

```python
missing_values = df.isnull().sum()
print(missing_values) 
for column in df.select_dtypes(include="number"):
    df[column] = df[column].fillna(df[column].mean())
```

```python
df.isnull().sum()
```

```python
top_10 = df.nlargest(10, "Happiness Score")

top_10[[
    "Country",
    "Economy (GDP per Capita)",
    "Health (Life Expectancy)"
]]
```

```python
import plotly.express as px

fig1 = px.bar(
    top_10,
    x="Country",
    y="Economy (GDP per Capita)",
    color="Health (Life Expectancy)",
    title="GDP per Capita and Healthy Life Expectancy of Top 10 Countries"
)

fig1.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt


columns = [
    "Economy (GDP per Capita)",
    "Family",
    "Health (Life Expectancy)",
    "Freedom",
    "Trust (Government Corruption)",
    "Generosity",
    "Happiness Score"
]

correlation = df[columns].corr()

plt.figure(figsize=(10, 7))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Between Happiness Factors")
plt.tight_layout()
plt.show()
```

```python
fig2 = px.scatter(
    df,
    x="Economy (GDP per Capita)",
    y="Happiness Score",
    color="Region",
    hover_name="Country",
    title="GDP per Capita vs Happiness Score"
)

fig2.show()
```

```python
region_scores = df.groupby("Region")["Happiness Score"].mean().reset_index()

fig3 = px.pie(
    region_scores,
    values="Happiness Score",
    names="Region",
    title="Average Happiness Score by Region"
)

fig3.show()
```

```python
fig4 = px.choropleth(
    df,
    locations="Country",
    locationmode="country names",
    color="Economy (GDP per Capita)",
    hover_name="Country",
    hover_data=["Health (Life Expectancy)"],
    title="GDP per Capita by Country"
)

fig4.show()
```

```python
import plotly.io as pio

graph1 = pio.to_html(
    fig1,
    full_html=False,
    include_plotlyjs="cdn"
)

graph2 = pio.to_html(
    fig2,
    full_html=False,
    include_plotlyjs=False
)

graph3 = pio.to_html(
    fig3,
    full_html=False,
    include_plotlyjs=False
)

graph4 = pio.to_html(
    fig4,
    full_html=False,
    include_plotlyjs=False
)


html = f"""
<html>
<head>
    <title>Happiness Data Analysis</title>
</head>

<body>

    <h1>Happiness Data Analysis</h1>

    <h2>1. GDP per Capita and Healthy Life Expectancy of Top 10 Countries</h2>
    {graph1}

    <h2>2. GDP per Capita vs Happiness Score</h2>
    {graph2}

    <h2>3. Average Happiness Score by Region</h2>
    {graph3}

    <h2>4. GDP per Capita by Country</h2>
    {graph4}

</body>
</html>
"""


with open("Happiness_Analysis.html", "w", encoding="utf-8") as file:
    file.write(html)

print("HTML page created successfully!")
```
